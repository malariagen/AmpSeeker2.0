```python
import json
```

```python
wkdir = ""
panel = 'ag-vampir'
```

```python
def add_remove_input_tag(input_nb, output_nb, cell_idx_to_change=2):

    with open(input_nb, mode='r', encoding='utf-8') as f:
        my_nb = json.loads(f.read())

    # Find the first code cell that is NOT tagged 'parameters' and NOT injected by papermill
    # to add remove-input. Use the original hardcoded index as fallback.
    target_idx = cell_idx_to_change
    for i, cell in enumerate(my_nb['cells']):
        tags = cell.get('metadata', {}).get('tags', [])
        src = ''.join(cell.get('source', []))
        # Skip papermill injected cell
        if 'injected-parameters' in tags:
            continue
        # Find first code cell with imports (no parameters tag)
        if cell.get('cell_type') == 'code' and 'parameters' not in tags and i > 0:
            # Check it looks like an import cell
            if 'import' in src or len(src.strip()) == 0:
                target_idx = i
                break

    if 'metadata' not in my_nb['cells'][target_idx]:
        my_nb['cells'][target_idx]['metadata'] = {}
    if 'tags' not in my_nb['cells'][target_idx]['metadata']:
        my_nb['cells'][target_idx]['metadata']['tags'] = []
    tags = my_nb['cells'][target_idx]['metadata']['tags']
    if 'remove-input' not in tags:
        tags.append('remove-input')

    with open(output_nb, 'w') as f:
        json.dump(my_nb, f)
```

## Add headers

```python
import glob

input_nbs = glob.glob(f"{wkdir}/docs/ampseeker-results/notebooks/*ipynb")
if panel == 'ag-vampir':
    other_nbs = glob.glob(f"{wkdir}/docs/ampseeker-results/notebooks/ag-vampir/*ipynb")
    input_nbs.extend(other_nbs)
```

```python
for nb in input_nbs:
    add_remove_input_tag(
    input_nb = nb,
    output_nb = nb,
)
```
